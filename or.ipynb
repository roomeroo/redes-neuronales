```python
class Neurona():
  def __init__(self, x1, x2) -> None:
    self.x1 = x1
    self.x2 = x2
    self.peso_x1 = 0.1
    self.peso_x2 = 0.2
    self.sesgo= 0.2
    self.tasa_aprendizaje = 0.1

  def predecir(self) -> float:
    z = self.x1 * self.peso_x1 +  self.x2 * self.peso_x2 + self.sesgo
    return (1 if z >= 0 else 0)

  def cambiar_sesgo(self, error:int):
    self.sesgo = self.tasa_aprendizaje * error
  def aprender(self,correcto)->None:
    prediccion = self.predecir()
    if(correcto!=prediccion):
      self.cambiar_sesgo(error=correcto-prediccion)
    print(self)

  def __str__(self) -> str:
    return f"sesgo: {self.sesgo:.2f}, x1: {self.x1}, x2: {self.x2} prediccion: {self.predecir()}"



```

```python
neurona1 = Neurona(0,0) # Lo correcto seria 0
neurona2 = Neurona(0,1) # Lo correcto seria 1
neurona3 = Neurona(1,0) #Lo correcto seria 1
neurona4 = Neurona(1,1) # Lo correcot seria 1
```

```python
print(neurona1.aprender(0))
print(neurona2.aprender(1))
print(neurona3.aprender(1))
print(neurona4.aprender(1))
```
